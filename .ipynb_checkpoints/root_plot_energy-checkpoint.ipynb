```python
import ROOT
```

```python
import matplotlib.pyplot as plt
```

```python
path = '~/Desktop/Privitera_335/panaSKImg_output/Am241-Spectra-data/avg_img_CV_20x630x500_bin10x10_125_20260228_072623_0.fz/'
files = [f'panaSKImg_clustersRec_avg_img_CV_20x630x500_bin10x10_125_20260228_072623_0_ext{i}.root' for i in [1,2,3,4]]
```

```python
# Create a TChain and specify the name of the TTree common to all files
chain = ROOT.TChain("clustersRec") # Replace "physics" with your TTree name
for file in files:
    chain.Add(path+file)
hist_combined = ROOT.TH1F("hist_combined", "Combined Energy Distribution;Energy (keV);Entries", 100, 0.5, 110)
chain.Project("hist_combined", "Energy", "")
c1 = ROOT.TCanvas("c1", "Combined Energy", 800, 600)
hist_combined.Draw()
c1.Update()
c1.Draw()
```

```python
c1.Close()
```

```python
hstack = ROOT.THStack("hs", "Combined Energy;Energy (keV);Entries")
legend = ROOT.TLegend(0.7, 0.7, 0.9, 0.9) # Add a legend

# Iterate through each file
for i, file in enumerate(files):
    f = ROOT.TFile.Open(path+file, "READ")
    tree = f.Get("clustersRec")

    # Create a unique histogram for this file
    hist_name = f"hist_{i}"
    hist = ROOT.TH1F(hist_name, "", 100, 0, 200)

    # Fill the histogram
    tree.Project(hist_name, "Energy", "")

    # Customize the histogram appearance
    hist.SetLineColor(i + 1) # Set a different color for each file (careful with color limits)
    hist.SetLineWidth(2)
    hist.SetFillStyle(3004) # Optional: adds a fill pattern
    hist.SetFillColor(i + 1)

    # Add to the THStack and Legend
    hstack.Add(hist)
    legend.AddEntry(hist, f"File {i+1}", "f")

    # Keep a reference to the histogram to prevent Python's garbage collector from deleting it prematurely
    globals()[hist_name] = hist

# Draw the THStack
c1 = ROOT.TCanvas("c1", "Stacked Energies", 800, 600)
# Use "nostack" option to overlay them instead of stacking them cumulatively
hstack.Draw("nostack hist")
legend.Draw()
c1.Update()
```
